```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Name' :['john' , 'adi' , 'noni'],
    'Age' : [20 , 21 , 22],
    'City' : ['New york' , 'Paris' , 'France'],
    'Salary' : [65000 , 78000 , 70000]

}
pd.DataFrame(data)
```

```python
data_list = [
    ['john' , 'adi' , 'noni'],
    [20 , 21 , 22],
    ['New york' , 'Paris' , 'France'],
    [65000 , 78000 , 70000]

]
df2 = pd.DataFrame(data_list)
columns = ["Name", "Age", "City", "Salary"]
df2 = pd.DataFrame(data_list).T
df2.columns = columns
df2
```
