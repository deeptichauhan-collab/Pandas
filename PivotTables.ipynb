```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Date" : pd.date_range("2020-01-01", periods=20),
    "Product" : ["A" , "B" , "C" , "D"]*5,
    "Region" : ["east" , "west" , "north" , "south"]*5,
    "Sales" : np.random.randint(100 , 1000 , 20),
    "Units" : np.random.randint(10 , 100 , 20),
    "Rep" : ["John" , "Alice" , "Mary" , "Alice"]*5

}
df = pd.DataFrame(data)
df["Month"] = df["Date"].dt.month_name()
df["Quarter"] = '0' + df["Date"].dt.quarter.astype(str)
df
```

```python
pd.pivot_table(df , values="Sales", index="Region", columns="Product" , aggfunc=np.sum)
```

```python
pivot2 = pd.pivot_table(df , values=["Sales" , "Units"] , index = "Region" , columns = "Product")
pivot2
```
