```python
import numpy as np
import pandas as pd
```

Merging Two Dataframes

```python
employees = pd.DataFrame({
    "employee_id" : [1 , 2 , 3 ],
    "name" : ["Deepti" , "Aditya" , "Ishita"],
    "department" : ["CSE" , "BCA" , "HR" ]
})
salaries = pd.DataFrame({
    "employee_id" : [1 , 2 , 3 ],
    "salary" : [20000 , 30000 , 40000],
    "bonus" : [5000 , 10000 , 40000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees , salaries, on="employee_id", how = "inner")
```

```python
pd.merge(employees , salaries, on="employee_id", how = "outer")
```

```python
pd.merge(employees , salaries, on="employee_id", how = "left")
```

CONCATENATION

```python
df1 = pd.DataFrame({
    "A" : ["A0" , "A1" , "A2" ],
    "B" : ["B0" , "B1" , "B2" ],
    "C" : ["C0" , "C1" , "C2" ]
})
df2 = pd.DataFrame({
    "A" : ["A3" , "A4" , "A5" ],
    "B" : ["B3" , "B4" , "B5" ],
    "C" : ["C3" , "C4" , "C5" ]
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1 , df2])
```

```python
pd.concat([df1 , df2] , axis = 1)
```

JOINING DATAFRAMES

```python
df1 = pd.DataFrame({
    "name" : ["Deepti" , "Aditya" , "Ishita"]
} , index = [1 , 2 , 3])

df2 = pd.DataFrame({
    "score" : [67 , 80 , 75]
} , index = [2 , 3 , 4])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2 , how = "outer")
```
